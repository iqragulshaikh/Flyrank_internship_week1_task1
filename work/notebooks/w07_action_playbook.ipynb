Action Queue Description:
The action queue prioritizes pages requiring immediate updates based on composite staleness and traffic volume scores.  

Reason Codes:

STALE_HIGH_TRAFFIC: High volume content with decaying engagement and high staleness score.   

CTR_DROP_STALE: Click-through rate falls below position expectations due to content age.   

LOW_IMPRESSION_MONITOR: Low volume or minor freshness degradation requiring passive tracking.

```python
import os
import pandas as pd
import numpy as np

# Generate sample ranked action queue for the playbook
np.random.seed(42)
n = 50
df = pd.DataFrame({
    'url': [f'https://example.com/page-{i}' for i in range(1, n+1)],
    'staleness_score': np.random.uniform(0.1, 0.99, n),
    'volume_score': np.random.uniform(0.1, 0.99, n)
})

df['score'] = (df['staleness_score'] * 0.6) + (df['volume_score'] * 0.4)
df['reason_code'] = np.where(df['score'] > 0.7, 'STALE_HIGH_TRAFFIC',
                     np.where(df['score'] > 0.4, 'CTR_DROP_STALE', 'LOW_IMPRESSION_MONITOR'))
df['action_label'] = np.where(df['reason_code'] == 'LOW_IMPRESSION_MONITOR', 'MONITOR', 'REFRESH_CONTENT')

ranked_queue = df.sort_values(by='score', ascending=False).reset_index(drop=True)
print(f"Ranked action queue generated successfully with {len(ranked_queue)} items.")
```
